# Циклы for и while

Цикл повторяет действие для набора объектов. Вместо ручного расчёта каждой продажи посчитаем дневную выручку.

Запускайте ячейки сверху вниз. После примера решите практические задачи и запишите, почему получили такой результат.

## Что меняется при каждом проходе?

Три замера температуры записаны по порядку в квадратных скобках. Цикл позволяет выполнить одно действие с каждым замером, даже если значения повторяются.

```python
for temperature in [18, 20, 18]:
    print(temperature)
```

На первом проходе имя temperature получает 18, на следующем — 20, на последнем — снова 18. Команда с отступом выполняется каждый раз для текущего значения. Последнее число тоже печатается, потому что цикл обходит все три позиции.

## Как накопить дневную выручку?

За три заказа заплатили 300, 500 и 200 тенге. Вместо отдельных имён для промежуточных сумм будем обновлять один накопитель и видеть его после каждой оплаты.

```python
total = 0
for payment in [300, 500, 200]:
    total = total + payment
    print(total)
```

Начальное значение ноль задаётся один раз перед циклом. Правая часть присваивания использует предыдущую сумму, прибавляет очередную оплату и сохраняет результат обратно в total. Если переместить обнуление внутрь цикла, на экране будут только суммы отдельных оплат.

## Когда закончится обратный отсчёт?

Для короткого таймера заранее известно условие продолжения: пока осталось хотя бы одно деление, нужно вывести число. Здесь удобно явно хранить остаток и уменьшать его.

```python
left = 3
while left > 0:
    print(left)
    left = left - 1
print("Старт!")
```

Перед каждой итерацией while проверяет выражение left > 0. После печати единицы счётчик уменьшается до нуля, и следующая проверка завершает цикл. Финальный print стоит без отступа и выполняется один раз даже при начальном left = 0.

## Демонстрация

Одна итерация отвечает за одну продажу. Накопитель, созданный внутри цикла, каждый раз обнулит прошлый результат. Для среднего надо отдельно обработать count=0. Изменять список во время обхода рискованно, проще собирать новый список.

```python
sales = [1000, 0, 2500, 500]
total = 0
count = 0
for amount in sales:
    total += amount
    if amount > 0:
        count += 1
print(total, count)
assert total == 4000
```
