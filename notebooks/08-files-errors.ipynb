## Как читать текст по одной строке?

Для опыта используем StringIO из стандартной библиотеки. Такой поток хранит текст в памяти и даёт потренироваться в чтении без создания файла на диске.

```python
from io import StringIO
with StringIO("Актау\nТараз\n") as source:
    for line in source:
        print(line.strip())
```

Последовательность символов обратная косая черта и n в строковом литерале обозначает перевод строки. Цикл получает строки по одной, а strip убирает пробельные символы по краям, включая перевод строки. Блок with закроет поток после выхода; для реального файла возможен такой же обход внутри with open(...).

## Как отличить плохую запись от числа ноль?

Из текстового файла поступили три цены. Одна запись содержит слово, поэтому перевод в целое число завершится исключением. Посчитаем общую сумму и отдельно число непригодных строк.

```python
from io import StringIO
total = 0
bad = 0
with StringIO("120\nошибка\n0\n") as source:
    for line in source:
        try:
            value = int(line.strip())
        except ValueError:
            bad += 1
            continue
        total += value
print(total, bad)
```

Поток StringIO имитирует файл в памяти. Блок try охватывает преобразование строки, а except ValueError обрабатывает случай текста, который нельзя прочитать как целое число. Команда continue пропускает сложение только для плохой строки, поэтому запись 0 остаётся корректной ценой.

## Как прочитать CSV с запятой внутри названия?

В CSV название товара иногда само содержит запятую. Хранение текста в памяти через StringIO позволяет проверить разбор одной строки с кавычками, прежде чем использовать настоящий файл.

```python
import csv
from io import StringIO
text = 'item,price\n"Чай, зелёный",450\n'
with StringIO(text) as source:
    for row in csv.DictReader(source):
        print(row["item"])
        print(int(row["price"]) * 2)
```

DictReader использует первую строку как названия полей и распознаёт запятую между кавычками как часть названия товара. Значение цены при чтении остаётся строкой, поэтому перед умножением на количество вызываем int. Для чтения того же формата с диска передайте DictReader файловый объект из open с параметрами encoding="utf-8" и newline="".

```python
# Создаём локальную учебную копию CSV для практики.
from pathlib import Path
DATA = Path("braincode_practice")
DATA.mkdir(exist_ok=True)
if not (DATA / "sales.csv").exists():
    (DATA / "sales.csv").write_text('item,quantity,price\nТетрадь,2,100\nРучка,3,50\nОшибка,1,bad\n', encoding="utf-8")
```

## Демонстрация

Режим w удаляет прежнее содержимое при открытии. В csv.DictReader все прочитанные значения сначала являются строками. Надо преобразовать цену в число и проверить результат. Нельзя молча ловить все исключения, иначе повреждённые данные выглядят как успешный запуск.

```python
import json
from pathlib import Path
path = Path("student.json")
student = {"name": "Алия", "score": 85}
with path.open("w", encoding="utf-8") as f:
    json.dump(student, f, ensure_ascii=False)
with path.open(encoding="utf-8") as f:
    loaded = json.load(f)
print(loaded)
assert loaded == student
```
