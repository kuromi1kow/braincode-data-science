# Копицентр: посчитать чек и сохранить код заявки

Копицентр присылает параметры заказа текстом. По трём полям подготовьте чек на печать и добавьте упаковку. Отдельно проверьте, что номер заявки в чеке выглядит так же, как на квитанции клиента.

В заявке указаны код '007', количество листов '8' и цена листа '12.50' в тенге. Упаковка за весь заказ стоит 20 тенге. Считайте цену неизменной для любого количества листов в этой задаче.

Выполняйте ячейки по порядку. Перед запуском попробуйте предсказать вывод.

## Шаг 1. Каким полям понадобятся преобразования перед расчётом?

```python
ticket = "007"
sheets_text = "8"
price_text = "12.50"
sheets = int(sheets_text)
price = float(price_text)
print(ticket, sheets, price)
```

**Обсуждение.** Количество листов целое, цена допускает дробную часть. Конечный ноль в записи цены исчез при преобразовании в число, но сама величина сохранилась. У кода заявки своя задача: последовательность символов нужна для сверки с оригиналом.

## Шаг 2. Сколько стоит печать и какой полный итог заказа?

```python
printing = sheets * price
packing = 20
total = printing + packing
print("Печать:", printing)
print("К оплате:", total)
```

**Обсуждение.** Восемь листов по 12,5 тенге дают 100 тенге. Упаковка приходится на весь заказ один раз. Чтобы вручную проверить выражение, подпишите единицы: листы при умножении на тенге за лист сокращаются, и оба слагаемых итоговой суммы выражены в тенге.

## Шаг 3. Как напечатать чек с полным кодом и двумя знаками после точки?

```python
print("Заявка", ticket)
print(f"Печать {sheets} листов: {printing:.2f} ₸")
print(f"Итого с упаковкой: {total:.2f} ₸")
```

**Обсуждение.** Внутри f-строки фигурные скобки подставляют значение переменной, а запись .2f задаёт вид выводимого числа. Здесь два знака нужны для единообразного чека. Значение total остаётся числом, пригодным для следующего расчёта.

## Поиск логической ошибки
Почему результат может ввести в заблуждение?

```python
print("Заявка", int(ticket))
print("К оплате:", sheets * (price + packing))
```
