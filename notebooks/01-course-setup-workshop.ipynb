# Пункт проката: насколько часто возвращают книги вовремя?

Пункт проката попросил оценить, сработает ли напоминание о возврате. Пока у нас есть только журнал за два дня до рассылки. Сначала выясним, какую долю возвращённых книг принесли вовремя. Затем решим, что из этого можно рассказать о напоминаниях.

Во вторник вернули 8 книг, из них 6 вовремя. В среду вернули 2 книги, из них 1 вовремя. Каждая книга входит в журнал один раз, в фактический день возврата. Это учебные данные. Все три шага выполняйте сверху вниз в одном блокноте.

Выполняйте ячейки по порядку. Перед запуском попробуйте предсказать вывод.

## Шаг 1. Сколько книг вернули с опозданием в каждый день?

```python
tue_returns = 8
tue_on_time = 6
wed_returns = 2
wed_on_time = 1
print(tue_returns - tue_on_time, wed_returns - wed_on_time)
```

**Обсуждение.** В журнале у вернувшейся книги один из двух исходов. Если из всех возвратов за день вычесть своевременные, останутся запоздалые. Быстрая сверка по дням помогает заметить ошибку ввода, например отрицательный остаток.

## Шаг 2. Какую долю всех возвратов за эти два дня принесли вовремя?

```python
returned = tue_returns + wed_returns
on_time = tue_on_time + wed_on_time
print(on_time, "из", returned)
print(100 * on_time / returned, "%")
```

**Обсуждение.** Нас интересует отдельная возвращённая книга, поэтому у всех десяти книг одинаковый вес. У семи из них отметка «вовремя». Читателю нужны оба числа рядом с процентом: выборки из десяти и тысячи возвратов дают разное представление об объёме наблюдений.

## Шаг 3. Какой вклад в итог внесли запоздалые возвраты и что именно измерил отчёт?

```python
late = returned - on_time
print("С опозданием:", late, "из", returned)
print("Доля опозданий:", 100 * late / returned, "%")
```

**Обсуждение.** Три опоздания можно сверить с суммой 2 + 1 на первом шаге. Значения 70 % и 30 % описывают только вернувшиеся книги. По этому журналу нельзя узнать, сколько книг до сих пор на руках или как повлияла ещё не начатая рассылка.

## Поиск логической ошибки
Почему результат может ввести в заблуждение?

```python
tue_percent = 100 * tue_on_time / tue_returns
wed_percent = 100 * wed_on_time / wed_returns
print((tue_percent + wed_percent) / 2, "%")
```

Среднее процентов даёт обоим дням равный вес, хотя во вторник наблюдений в четыре раза больше. Это ответ о среднем дневном показателе. Для доли книг за весь период используйте суммарные количества из второго шага.
