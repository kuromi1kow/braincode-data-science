## Как дать числу понятное имя?

На складе осталось семь коробок. Сохраните это число под именем boxes, чтобы использовать его в следующей строке без повторного ввода.

```python
boxes = 7
print(boxes * 4)
```

Знак равенства связывает имя слева со значением справа. Во второй строке Python подставляет вместо boxes число 7. Если в каждой коробке четыре кружки, произведение даёт общее количество кружек.

## Почему «12» и 12 ведут себя по-разному?

При чтении формы количество часто приходит как текст. Сначала посмотрите, как Python соединяет текст, затем переведите тот же ввод в целое число.

```python
entered = "12"
print(entered + "3")
print(int(entered) + 3)
```

Кавычки делают 12 строкой, и плюс между двумя строками склеивает их символы. Вызов int(entered) возвращает целое число 12. Теперь по обе стороны плюса числа, поэтому последняя строка печатает сумму 15.

## Сколько получится полных упаковок?

В отделе выдачи нужно разложить 29 маркеров по коробкам по шесть. Одного деления мало: нам нужны и количество заполненных коробок, и число оставшихся маркеров.

```python
markers = 29
capacity = 6
full = markers // capacity
left = markers % capacity
print(full, left)
```

Оператор // для этих положительных чисел даёт целую часть частного, а % даёт остаток. Имена full и left хранят разные ответы на один вопрос о делении. Проверить вывод можно обратным действием: четыре коробки по шесть плюс пять маркеров дают исходные 29.

## Демонстрация

Вычисление выполняется справа от знака присваивания, затем имя связывается с результатом. Формат .2f влияет на вывод, а не точность хранения. Для денежных расчётов в продукте обычно применяют целые минимальные единицы или Decimal.

```python
price = 2500
quantity = 3
discount = 0.10
subtotal = price * quantity
total = subtotal * (1 - discount)
print(type(price).__name__)
print(f"К оплате: {total:.2f} ₸")
assert total == 6750
```
