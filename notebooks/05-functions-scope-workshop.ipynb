# Сколько продлится выпуск студенческого подкаста?

Редактор соединяет две записи и вставляет заставку между ними. Подготовьте функции для перевода длительностей в общую единицу, вычисления итога и печати в формате минут и секунд.

Первый фрагмент длится 1 минуту 59 секунд, второй — 1 минуту 5 секунд, заставка — 12 секунд. В исходных парах секундная часть лежит в диапазоне 0–59.

Выполняйте ячейки по порядку. Перед запуском попробуйте предсказать вывод.

## Шаг 1. Как сравнивать и складывать длительности?

```python
def to_seconds(minutes, seconds):
    return minutes * 60 + seconds

first = to_seconds(1, 59)
second = to_seconds(1, 5)
print(first, second)
```

**Обсуждение.** У каждой исходной записи два числа в разных единицах. Функция переводит минуты в секунды и возвращает одно целое число. После вызовов first и second содержат сопоставимые величины.

## Шаг 2. Как добавить заставку один раз на стыке фрагментов?

```python
def episode_length(left, right, intro):
    return left + intro + right

length = episode_length(first, second, 12)
print(length)
```

**Обсуждение.** Параметры left, right и intro обозначают секунды и получают значения из вызова. Внутри функции заставка участвует в сложении ровно один раз. Результат можно сохранить в length и передать дальше, поэтому внутри функции ничего печатать не требуется.

## Шаг 3. Как вернуть слушателю привычную запись времени?

```python
def time_label(total):
    minutes = total // 60
    seconds = total % 60
    return f"{minutes}:{seconds:02d}"

print(time_label(length))
print(time_label(65))
```

**Обсуждение.** Целочисленное деление даёт число полных минут, а остаток — секунды после них. В форматированной строке выражение seconds:02d отводит под целое число две позиции и при необходимости ставит слева ноль. Отдельная проверка числа 65 помогает заметить пропущенный ведущий ноль.

## Поиск логической ошибки
Почему результат может ввести в заблуждение?

```python
as_digits = 159 + 105 + 12
print(as_digits // 100, as_digits % 100)
```

Записи 1:59 и 1:05 были склеены в десятичные числа. Цифры переносятся в следующий разряд после ста, а минуты должны увеличиваться после шестидесяти секунд. Поэтому 2 минуты 76 секунд — некорректная запись; используйте to_seconds перед сложением и time_label после него.
