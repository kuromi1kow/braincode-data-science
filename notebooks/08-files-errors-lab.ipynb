### Слайд 3. Путь к файлу

Path соединяет имя папки с именем файла.

```python
from pathlib import Path

folder = Path("braincode_08")
folder.mkdir(exist_ok=True)
path = folder / "notes.txt"
print(path.name)
```

### Слайд 4. Запись текста

Режим w записывает новый текст. encoding сохраняет кириллицу.

```python
with path.open("w", encoding="utf-8") as file:
    file.write("Чай\n")
print("Записано")
```

### Слайд 5. Чтение текста

read() возвращает содержимое файла в виде строки.

```python
with path.open("r", encoding="utf-8") as file:
    text = file.read()
print(text.rstrip("\n"))
```

### Слайд 7. Добавление строки

Режим a добавляет данные после уже записанного текста.

```python
with path.open("a", encoding="utf-8") as file:
    file.write("Кофе\n")

print(path.read_text(encoding="utf-8").rstrip())
```

### Слайд 8. Когда файла нет

FileNotFoundError возникает при попытке прочитать отсутствующий файл.

```python
missing = folder / "no_such_file.txt"
try:
    missing.read_text(encoding="utf-8")
except FileNotFoundError:
    print("Файл не найден")
```

### Слайд 9. Когда данные нельзя преобразовать

int("bad") вызывает ValueError.

```python
for value in ["250", "bad"]:
    try:
        print(int(value))
    except ValueError:
        print("Некорректное число", value)
```

### Слайд 11. Сохранение JSON

json.dump записывает словарь в открытый файл.

```python
import json

item = {"name": "Чай", "price": 250}
info = folder / "item.json"
with info.open("w", encoding="utf-8") as file:
    json.dump(item, file, ensure_ascii=False)
print(info.name)
```

### Слайд 12. Чтение JSON

json.load возвращает данные, сохранённые в JSON.

```python
with info.open("r", encoding="utf-8") as file:
    item = json.load(file)
print(item["name"], item["price"])
```

### Слайд 14. Чтение строк CSV

DictReader использует заголовки как ключи словаря.

```python
import csv

sales = folder / "sales.csv"
with sales.open(encoding="utf-8", newline="") as f:
    reader = csv.DictReader(f)
    row = next(reader)
print(row["item"], type(row["quantity"]).__name__)
```

### Слайд 15. Выручка одной строки

Преобразуем цену и количество в числа перед умножением.

```python
quantity = int(row["quantity"])
price = int(row["price"])
revenue = quantity * price
print(revenue)
```

### Слайд 16. Сложение корректных продаж

Ошибочную строку запоминаем, затем продолжаем перебор.

```python
total = 0
rejected = []
with sales.open(encoding="utf-8", newline="") as f:
    for row in csv.DictReader(f):
        try:
            quantity = int(row["quantity"])
            total += quantity * int(row["price"])
        except ValueError:
            rejected.append(row["item"])
print(total)
print(rejected)
```

### Слайд 17. Сохранение отчёта

Итог и список ошибок можно сохранить вместе в JSON.

```python
report = {"revenue": total, "rejected": rejected}
summary = folder / "summary.json"
with summary.open("w", encoding="utf-8") as file:
    json.dump(
        report, file, ensure_ascii=False, indent=2
    )
print(summary.name)
```
