### Слайд 2. Знакомые функции

Вызов функции записывают с круглыми скобками.

```python
sales = [100, 200, 300]
count = len(sales)
print(count)
```

### Слайд 4. Определение и вызов

def создаёт функцию. Вызов запускает её тело.

```python
def greet():
    print("Привет!")

greet()
```

### Слайд 5. Повторный вызов

Каждый вызов выполняет тело функции ещё раз.

```python
def greet():
    print("Привет!")

greet()
greet()
```

### Слайд 6. Параметр и аргумент

name получает значение, переданное при вызове.

```python
def greet(name):
    print("Привет,", name)

greet("Мира")
greet("Алия")
```

### Слайд 7. Два параметра

При вызове по порядку параметр price получает первый аргумент.

```python
def order_cost(price, quantity):
    return price * quantity

print(order_cost(150, 3))
```

### Слайд 8. Что делает return

return передаёт результат в место вызова.

```python
def order_cost(price, quantity):
    return price * quantity

total = order_cost(150, 3)
print(total + 50)
```

### Слайд 9. print() и return

Без явного return функция возвращает None.

```python
def show_total():
    print(450)

result = show_total()
print(result)
```

### Слайд 10. Порядок и имена аргументов

Именованный аргумент явно указывает параметр.

```python
def order_cost(price, quantity):
    return price * quantity

print(order_cost(150, 3))
print(order_cost(quantity=3, price=150))
```

### Слайд 11. Значение по умолчанию

Если delivery пропустить, функция возьмёт 0.

```python
def order_total(total, delivery=0):
    return total + delivery

print(order_total(450))
print(order_total(450, 50))
```

### Слайд 14. Локальная переменная

Имена параметров и subtotal доступны внутри функции.

```python
def order_cost(price, quantity):
    subtotal = price * quantity
    return subtotal

total = order_cost(150, 3)
print(total)
```

### Слайд 15. Одинаковое имя внутри и снаружи

Присваивание в теле создаёт локальное имя score.

```python
score = 10

def local_score():
    score = 20
    return score

print(local_score())
print(score)
```

### Слайд 16. Условие внутри функции

Когда выполняется return, вызов заканчивается.

```python
def delivery_cost(total):
    if total >= 5000:
        return 0
    return 500

print(delivery_cost(4999))
print(delivery_cost(5000))
```

### Слайд 17. Цикл внутри функции

Функция получает список и возвращает сумму.

```python
def total_sales(sales):
    total = 0
    for amount in sales:
        total = total + amount
    return total

print(total_sales([100, 200, 300]))
print(total_sales([]))
```

### Слайд 18. Проверка пустого списка

Пустой список проверяем перед делением.

```python
def average_or_none(values):
    if len(values) == 0:
        return None
    return sum(values) / len(values)

print(average_or_none([10, 20]))
print(average_or_none([]))
```

### Слайд 19. Вызов одной функции внутри другой

Готовое правило доставки можно использовать повторно.

```python
def delivery_cost(total):
    if total >= 5000:
        return 0
    return 500

def order_total(total):
    return total + delivery_cost(total)

print(order_total(4500))
```
