# Проверить режим учебного датчика на всех порогах

В учебном макете табло показывает один режим в зависимости от числа занятых мест. Реализуйте правило, проверьте вход слева и справа от порога, а затем поймайте ошибку с порядком веток. Вместимость зала — десять мест.

Датчик возвращает целое число occupied. Отрицательное значение и число больше десяти обозначают ошибку. Если свободных мест три и больше, табло пишет «Можно входить». При одном или двух свободных — «Осталось мало мест». При нуле свободных — «Зал полон».

Выполняйте ячейки по порядку. Перед запуском попробуйте предсказать вывод.

## Шаг 1. Сколько свободных мест и допустимо ли показание 8?

```python
capacity = 10
occupied = 8
free = capacity - occupied
valid = 0 <= occupied <= capacity
print("Свободно:", free)
print("Допустимо:", valid)
```

**Обсуждение.** Правило табло описано через свободные места, а датчик передаёт занятые. Вычитание связывает две величины. Проверка допустимости понадобится до выбора сообщения: даже неверное число можно вычесть из десяти и получить внешне убедительный результат.

## Шаг 2. Как выбрать единственное сообщение для этих двух свободных мест?

```python
if not valid:
    message = "Ошибка датчика"
elif free == 0:
    message = "Зал полон"
elif free <= 2:
    message = "Осталось мало мест"
else:
    message = "Можно входить"
print(message)
```

**Обсуждение.** Сначала цепочка отводит ошибочные показания в отдельную ветку. До проверки free <= 2 доходят только допустимые положительные остатки, потому что случай free == 0 проверен раньше. Поэтому в третью ветку попадут только одно и два свободных места.

## Шаг 3. Как убедиться, что на соседнем значении сообщение действительно поменяется?

```python
occupied = 7
free = capacity - occupied
valid = 0 <= occupied <= capacity
if not valid:
    message = "Ошибка датчика"
elif free == 0:
    message = "Зал полон"
elif free <= 2:
    message = "Осталось мало мест"
else:
    message = "Можно входить"
print(free, message)
```

**Обсуждение.** После изменения occupied заново вычислите зависимые величины. Прежнее значение free осталось бы равным двум и скрывало бы переход. Соседние показания 7 и 8 проверяют именно границу между двумя сообщениями при неизменной вместимости.
