## Попадает ли заказ на границу тарифа?

Бесплатная доставка начинается с 5 000 тенге включительно. Перед тем как писать весь тариф, проверим одно сравнение ровно на пороге.

```python
amount = 5000
print(amount >= 5000)
print(amount > 5000)
```

Оператор >= включает равное значение, поэтому первое выражение даёт True — истину. Строгий знак > проверяет только превышение, и второе выражение даёт False — ложь. Такое различие легко пропустить, если проверять только заказы на 6 000.

## Как программа выбирает одно сообщение?

Посылку можно забрать, если в ячейке осталась хотя бы одна коробка. Подставим ноль и проследим, какая часть программы выполнится.

```python
boxes = 0
if boxes > 0:
    print("Можно забирать")
else:
    print("Ячейка пуста")
```

Строка if задаёт проверку, а отступ в следующей строке показывает, какое действие зависит от неё. Ноль не больше нуля, поэтому Python сразу переходит к ветке else. Если изменить boxes на 1, будет выбран первый вывод.

## Почему важен порядок скидок?

При сумме от 10 000 участник клуба получает скидку 15 процентов. Для остальных заказов от 5 000 действует скидка 5 процентов. Запишем правила в порядке проверки.

```python
amount = 12000
member = True
if member and amount >= 10000:
    discount = 15
elif amount >= 5000:
    discount = 5
else:
    discount = 0
print(discount)
```

Оператор and требует, чтобы одновременно выполнялись оба условия первой строки проверки. Заказ участника на 12 000 подходит, и Python пропускает оставшиеся ветки цепочки. Если поставить более широкий порог 5 000 первым, этот же заказ преждевременно получит только пять процентов.

## Демонстрация

В цепочке if/elif выполняется только первая подходящая ветка. Если поставить amount >= 5000 раньше amount >= 10000, заказ на 12000 ошибочно получит платную доставку. Отступы задают принадлежность инструкции к блоку.

```python
amount = 5000
if amount < 0:
    raise ValueError("Сумма не может быть отрицательной")
elif amount >= 10000:
    delivery = 0
elif amount >= 5000:
    delivery = 500
else:
    delivery = 1000
print(delivery)
```
