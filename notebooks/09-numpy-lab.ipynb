### Слайд 3. Список и массив

Умножение массива на 2 меняет каждое число.

```python
import numpy as np
prices = [500, 350]
a = np.array(prices)

print(prices * 2)
print(a * 2)
```

### Слайд 5. Индексы и срезы

Первый элемент находится по индексу 0.

```python
a = np.array([500, 350, 250])
print(a[0])
print(a[0:2])
```

### Слайд 6. Сумма каждого заказа

Цена умножается на количество на той же позиции.

```python
qty = np.array([1, 2, 3])
revenue = a * qty
print(revenue)
```

### Слайд 7. Сумма и среднее

sum() и mean() дают по одному числу.

```python
print(revenue.sum())
print(revenue.mean())
```

### Слайд 8. Маска отбора

Сравнение применяется к каждому элементу.

```python
keep = revenue >= 700
print(keep)
```

### Слайд 11. Выбор элементов по маске

Массив оставляет значения на позициях с True.

```python
selected = revenue[keep]
print(selected)
print(selected.sum())
```

### Слайд 12. Массив из строк и столбцов

Две строки по три значения дают форму (2, 3).

```python
table = np.array([[1, 2, 3],
                  [4, 5, 6]])
print(table.shape)
print(table[1, 0])
```

### Слайд 13. Суммы по строкам и столбцам

Параметр axis выбирает направление свёртки.

```python
print(table.sum(axis=0))
print(table.sum(axis=1))
```

### Слайд 15. Прогноз выручки при новых ценах

Добавим 50 тенге к цене каждой единицы.

```python
new_prices = a + 50
new_revenue = new_prices * qty
print(revenue.sum())
print(new_revenue.sum())
```

### Слайд 18. Чтение чужого решения

Форма результата помогает понять, что именно посчитали.

```python
per_order = a * qty
selected = per_order[per_order >= 700]
print(selected.size)
```
