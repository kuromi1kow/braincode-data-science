## Как сохранить порядок ожидания?

В очереди сейчас Алия и Марат, затем приходит Дана. Список сохраняет порядок добавления и позволяет обратиться к человеку по позиции.

```python
queue = ["Алия", "Марат"]
queue.append("Дана")
print(queue[0])
print(len(queue))
```

Метод append добавляет один элемент в конец существующего списка. Индекс 0 обозначает первую позицию, а len считает все элементы. Поэтому первым по-прежнему остаётся прежний посетитель, хотя очередь стала длиннее.

## Как посчитать заказы по городу?

Журнал содержит названия городов по одному на заказ. Для каждого города нужен свой счётчик, поэтому используем словарь, в котором ключом станет название города.

```python
counts = {}
for city in ["Актау", "Тараз", "Актау"]:
    counts[city] = counts.get(city, 0) + 1
print(counts["Актау"], counts["Тараз"])
```

Пустые фигурные скобки создают словарь. Вызов get возвращает прежнее значение по ключу или начальный ноль, когда город встречается впервые. После прибавления единицы записываем счётчик под тем же ключом, поэтому второй заказ из Актау обновляет прежнюю запись.

## Сколько клиентов вернулись на следующий день?

История покупок может содержать много записей одного клиента. Для подсчёта людей оставим по одному идентификатору на день и выделим общих клиентов двух дней.

```python
monday = set(["A", "A", "B", "C"])
tuesday = {"B", "C", "D"}
returned = monday & tuesday
print(len(returned))
print(sorted(returned))
```

Множество set сохраняет уникальные элементы, так что два появления A дают одного человека. Операция & берёт пересечение, то есть идентификаторы, встречающиеся в обоих множествах. Порядок множества не задан, поэтому sorted выводит имена в устойчивом порядке для сверки.

## Демонстрация

Метод get возвращает значение по ключу или заданный запасной результат. Срез списка копирует контейнер, но вложенные объекты могут оставаться общими. tuple нельзя изменить по индексу, но он может содержать изменяемый список.

```python
clients = ["A", "B", "A", "C"]
counts = {}
for client in clients:
    counts[client] = counts.get(client, 0) + 1
unique_clients = set(clients)
print(counts)
print(len(unique_clients))
print(set(["A", "D"]) & unique_clients)
```
