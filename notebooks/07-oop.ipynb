Класс объединяет состояние и действия объекта. Это удобно, когда у нескольких объектов одинаковое поведение и независимые данные.

## Как объединить имя участника и его данные?

Представим, что в учебной системе есть карточки студентов. Для начала карточке достаточно одного имени. Опишем общий способ создания карточек через класс.

```python
class Student:
    def __init__(self, name):
        self.name = name

first = Student("Алия")
print(first.name)
```

Вызов Student создаёт экземпляр класса и передаёт имя методу __init__ для начальной настройки. Параметр self внутри метода указывает на создаваемый объект. Присваивание self.name сохраняет значение в объекте, поэтому его можно прочитать после завершения метода.

## Откуда метод берёт прошлое значение?

Посчитаем посещения одного участника. Храним текущее число внутри объекта и вызываем метод один раз на каждое посещение.

```python
class Visits:
    def __init__(self):
        self.count = 0
    def mark(self):
        self.count += 1

record = Visits()
record.mark()
record.mark()
print(record.count)
```

Метод __init__ устанавливает ноль при создании экземпляра. При вызове record.mark() Python передаёт этот же экземпляр в параметр self, а += 1 увеличивает его счётчик. Ко второму вызову первый результат сохраняется, поэтому получаем два.

## Могут ли две группы хранить разные списки?

Пусть один студент записался в утреннюю группу, а вечерняя ещё пуста. У каждой группы должен быть собственный список записей, иначе изменения случайно смешаются.

```python
class Group:
    def __init__(self):
        self.names = []
    def enroll(self, name):
        self.names.append(name)

morning = Group()
evening = Group()
morning.enroll("Дана")
print(morning.names, evening.names)
```

Выражение [] выполняется заново при каждом вызове __init__, создавая отдельный список. Метод enroll изменяет список того экземпляра, через который его вызвали. Пустой вывод для evening служит короткой проверкой независимости двух групп.

## Демонстрация

self указывает на конкретный экземпляр. total в __init__ относится к одному объекту. Изменяемый список на уровне класса стал бы общим для экземпляров. Для простых расчётов функция часто яснее класса. Scikit-learn также использует объекты с методами fit и predict.

```python
class Cart:
    def __init__(self):
        self.total = 0
    def add(self, price):
        if price < 0:
            raise ValueError("Отрицательная цена")
        self.total += price

a, b = Cart(), Cart()
a.add(100)
a.add(250)
print(a.total, b.total)
```
