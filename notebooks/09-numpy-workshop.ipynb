# Какие фасовочные линии выходят за допуск?

На фабрике три линии наполняют пакеты разного размера. Вам передали четыре раунда контрольного взвешивания. Вычислите отклонения от нормы, найдите нарушения и покажите, на какой линии проверку нужно повторить первой.

Строка матрицы — один раунд, столбец — одна линия. Цели по линиям равны 250, 500 и 1000 г. Допустимые отклонения в любую сторону составляют соответственно 3, 5 и 8 г. Точное попадание на границу считается нормой.

Выполняйте ячейки по порядку. Перед запуском попробуйте предсказать вывод.

## Шаг 1. Сколько граммов прибавила или недоложила каждая линия?

```python
import numpy as np

mass = np.array([[250, 496, 1010],
                 [247, 507, 1001],
                 [254, 500, 989],
                 [249, 505, 1008]])
target = np.array([250, 500, 1000])
tolerance = np.array([3, 5, 8])
delta = mass - target
print(delta.shape)
print(delta.tolist())
```

**Обсуждение.** Вектор target содержит одну норму на столбец. При вычитании NumPy применяет одни и те же три нормы к каждой строке. Например, третий пакет первой линии весит 254 г, значит отклонение равно +4 г. Знак сохраняет направление отклонения.

## Шаг 2. Какие измерения нарушают двусторонний допуск?

```python
outside = np.abs(delta) > tolerance
print(outside.astype(int).tolist())
print("Нарушений:", outside.sum())
```

**Обсуждение.** Абсолютное значение позволяет одинаково учитывать недовес и перевес. В выводе единица означает нарушение. В последней строке отклонения +5 и +8 г равны своим допускам, поэтому остаются в допустимом диапазоне.

## Шаг 3. Какая линия нарушала допуск чаще других и в каком раунде больше всего нарушений?

```python
by_line = outside.sum(axis=0)
by_round = outside.sum(axis=1)
print("По линиям:", by_line.tolist())
print("По раундам:", by_round.tolist())
print("Линия для проверки:", int(by_line.argmax()) + 1)
```

**Обсуждение.** Сумма по оси 0 убирает раунды и оставляет три результата по линиям. У третьей линии два выхода за допуск из четырёх взвешиваний, а третий раунд содержит два нарушения. argmax выдаёт индекс 2. В заводской нумерации это линия 3.

## Поиск логической ошибки
Почему результат может ввести в заблуждение?

```python
import numpy as np

faulty_delta = np.array([[0, -4, 10], [-3, 7, 1],
                         [4, 0, -11], [-1, 5, 8]])
limit = np.array([3, 5, 8])
print((faulty_delta > limit).sum(axis=0).tolist())
```

Сравнение со знаком обнаружило перевес, но пропустило −11 г на третьей линии. Этот пакет вышел за нижнюю границу 992 г и тоже требует внимания. Примените абсолютное значение перед сравнением или явно проверьте обе границы.
