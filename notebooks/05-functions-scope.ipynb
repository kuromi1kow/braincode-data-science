## Как повторить расчёт для двух заказов?

Стоимость строки заказа равна цене за единицу, умноженной на количество. Дадим этому правилу имя, чтобы его можно было вызвать с разными числами.

```python
def cost(price, count):
    return price * count

print(cost(250, 3))
print(cost(100, 2))
```

Строка def объявляет функцию, а price и count служат именами для входных значений. При первом вызове они получают 250 и 3. Команда return отдаёт произведение вызывающему коду, поэтому одно и то же тело обслуживает оба заказа.

## Где живёт промежуточная сумма?

При расчёте суммы с доставкой удобно сохранить промежуточное число. Даже если снаружи уже существует имя total, функция может иметь своё локальное значение.

```python
total = 50
def with_delivery(goods):
    total = goods + 200
    return total

print(with_delivery(600))
print(total)
```

Присваивание внутри функции создаёт локальное имя total. Вызов возвращает 800, а после него команда print(total) обращается к внешнему значению 50. Чтобы использовать результат позже, его можно сохранить при вызове под отдельным именем.

## Что вернуть, если наблюдений пока нет?

Для отчёта требуется средняя сумма продажи. До первого заказа среднее вычислить нельзя, поэтому договоримся возвращать специальное значение None для пустого набора.

```python
def average(total, count):
    if count == 0:
        return None
    return total / count

print(average(900, 3))
print(average(0, 0))
```

Условие проверяется до деления. При нулевом count первый return сразу завершает функцию, так что до опасной операции выполнение не дойдёт. Результат None помогает отличить отсутствие наблюдений от среднего ноль, которое бывает при реальных бесплатных покупках.

## Демонстрация

Отсутствующий return означает возврат None. Не надо изменять глобальную total ради вычисления среднего. Изменяемый аргумент по умолчанию, например values=[], создаётся один раз и может сохранять состояние между вызовами.

```python
def average(values):
    if not values:
        return None
    return sum(values) / len(values)

result = average([2, 4, 6])
print(result)
assert result == 4
assert average([]) is None
```
