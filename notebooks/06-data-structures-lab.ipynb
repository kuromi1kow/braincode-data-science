## Примеры с лекции

Запускайте каждый пример целиком. Под ним появится результат.

### Слайд 3. Список

Список хранит порядок и допускает повторения.

```python
sales = [100, 100, 250]
print(sales)
print(len(sales))
```

### Слайд 4. Индексы списка

Первый элемент имеет индекс 0.

```python
sales = [100, 250, 400]
print(sales[0])
print(sales[1])
print(sales[-1])
```

### Слайд 5. Изменение списка

Присваивание меняет элемент, append добавляет новый в конец.

```python
sales = [100, 250]
sales[1] = 300
sales.append(450)
print(sales)
```

### Слайд 6. Срез списка

Срез [1:3] берёт элементы с индексами 1 и 2.

```python
sales = [100, 250, 400, 500]
print(sales[1:3])
print(sales)
```

### Слайд 7. Копия списка

copy() создаёт отдельный список для показанных чисел.

```python
sales = [100, 250]
draft = sales.copy()
draft.append(400)
print(sales)
print(draft)
```

### Слайд 8. Кортеж

Кортеж хранит порядок. Его элементы нельзя заменить.

```python
point = (3, 5)
print(point[0])
print(point[1])
```

### Слайд 9. Распаковка кортежа

x получает первое значение, y получает второе.

```python
point = (3, 5)
x, y = point
print(x)
print(y)
```

### Слайд 10. Словарь

Словарь связывает ключи со значениями.

```python
item = {"name": "Чай", "price": 350}
print(item["name"])
print(item["price"])
```

### Слайд 11. Изменение словаря

Известный ключ обновляется, новый ключ добавляется.

```python
item = {"name": "Чай", "price": 350}
item["price"] = 400
item["stock"] = 8
print(item["price"])
print(item["stock"])
```

### Слайд 12. Ключ может отсутствовать

get() возвращает запасное значение, если ключа нет.

```python
item = {"price": 350}
print(item.get("price", 0))
print(item.get("stock", 0))
```

### Слайд 13. Перебор пар словаря

items() передаёт ключ и значение в каждом шаге.

```python
counts = {"Актау": 2, "Атырау": 1}
for city, count in counts.items():
    print(city, count)
```

### Слайд 16. Множество

set хранит уникальные элементы.

```python
cities = ["Актау", "Атырау", "Актау"]
unique = set(cities)
print(len(unique))
print(sorted(unique))
```

### Слайд 17. Проверка принадлежности

in отвечает, есть ли значение в множестве.

```python
cities = {"Актау", "Атырау"}
print("Актау" in cities)
print("Алматы" in cities)
```

### Слайд 18. Общие и все уникальные элементы

intersection ищет общие элементы, union собирает все уникальные.

```python
monday = {101, 102}
tuesday = {102, 103}
common = monday.intersection(tuesday)
all_ids = monday.union(tuesday)
print(sorted(common))
print(sorted(all_ids))
```

### Слайд 20. Список записей

Список хранит заказы, словарь описывает один заказ.

```python
orders = [
    {"city": "Актау", "amount": 100},
    {"city": "Атырау", "amount": 250},
]
for order in orders:
    print(order["city"], order["amount"])
```
