# Сколько книг доступно в каждом филиале?

Библиотеки передают повторные сообщения об остатке одного учебника. Каждая строка заменяет предыдущее состояние того же филиала. Сформируйте текущую сводку и поясните читателю разницу между нулём на полке и отсутствием сведений.

Сообщения расположены от старых к новым. Для отчёта нужны филиалы Юг, Центр, Север и Парк. Каждая запись содержит название филиала и целое неотрицательное число доступных экземпляров.

Выполняйте ячейки по порядку. Перед запуском попробуйте предсказать вывод.

## Шаг 1. Сколько раз обновили сведения и сколько филиалов отчитались?

```python
updates = [("Юг", 3), ("Центр", 2), ("Юг", 1),
           ("Север", 0), ("Центр", 4)]
branches = set()
for name, count in updates:
    branches.add(name)
print(len(updates), len(branches))
```

**Обсуждение.** Список хранит пять сообщений в порядке поступления. В каждой итерации пара распаковывается в название и количество. Множество собирает разные названия, поэтому повторные сообщения оставляют в нём всего три филиала.

## Шаг 2. Как выбрать для каждого филиала самый свежий остаток?

```python
latest = {}
for name, count in updates:
    latest[name] = count
print(latest["Юг"], latest["Центр"], latest["Север"])
```

**Обсуждение.** Словарь связывает один ключ с одним значением. Поскольку список упорядочен по времени, присваивание тому же ключу заменяет старые сведения свежими. Для Юга число 3 заменилось на 1, а для Центра 2 заменилось на 4.

## Шаг 3. Как показать отсутствие сведений отдельным сообщением?

```python
required = ["Юг", "Центр", "Север", "Парк"]
for name in required:
    if name in latest:
        print(name, latest[name])
    else:
        print(name, "нет сведений")
```

**Обсуждение.** Условие name in latest проверяет присутствие ключа в словаре. Ключ «Север» существует, поэтому число ноль попадает в отчёт. «Парк» отсутствует, и отдельная ветка сообщает об этом без догадки об остатке.

## Поиск логической ошибки
Почему результат может ввести в заблуждение?

```python
for name in ["Север", "Парк"]:
    if latest.get(name):
        print(name, latest[name])
    else:
        print(name, "нет сведений")
```

В условии ноль превращается в ложь, и отсутствующий ключ через get даёт None, который тоже ложен. Поэтому эти разные ситуации получили одинаковую подпись. В данном отчёте правильный вопрос — есть ли ключ в словаре; после проверки можно выводить любое значение, включая ноль.
