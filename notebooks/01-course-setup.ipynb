## Что происходит при запуске ячейки?

В блокноте можно вычислять результат и сразу выводить его на экран. Запустите две строки вместе и проследите, в каком порядке появятся ответы.

```python
print("Заказы за день")
print(12 + 8)
```

Команда print показывает то, что стоит в скобках. В первой строке кавычки обозначают текст, поэтому слова появятся на экране без кавычек. Во второй строке Python сначала сложит числа, а затем покажет сумму.

## Сколько стоит весь заказ?

Покупатель взял три бутылки воды по 200 тенге и один сэндвич за 650. Запишите расчёт одной строкой и заранее прикиньте, должна ли сумма превысить тысячу.

```python
print(3 * 200 + 650)
```

Звёздочка обозначает умножение. Оно выполняется раньше сложения, поэтому три бутылки дают 600, а сэндвич добавляет ещё 650. Это стоимость одного заказа в тенге.

## Какую долю посетителей мы посчитали?

Утром зашли 20 человек и купили товар четверо. Вечером зашли 80 человек и купили ещё шестеро. Найдём процент покупателей среди всех посетителей за день.

```python
print((4 + 6) / (20 + 80) * 100)
```

В первых скобках мы сложили покупателей, во вторых всех посетителей. Косая черта делит 10 на 100, затем умножение на 100 переводит долю в проценты. В выводе 10.0 обозначает десять процентов, а точка связана с тем, что деление вернуло дробное число.

## Демонстрация
Обе формулы допустимы и отвечают на разные вопросы. Вечером посетителей девять раз больше. Для вопроса о случайно выбранном визите нужен общий знаменатель 100. Поменяйте число вечерних посетителей и проследите, как изменятся ответы.

```python
morning = 5 / 10
evening = 9 / 90
print((morning + evening) / 2)
print((5 + 9) / (10 + 90))
```
