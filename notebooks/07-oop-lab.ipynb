### Слайд 4. Первый класс

Cart() создаёт экземпляр и запускает __init__ внутри него.

```python
class Cart:
    def __init__(self):
        self.total = 0

cart = Cart()
print(cart.total)
```

### Слайд 5. Аргумент при создании

При Student("Мира") параметр name получает строку.

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.scores = []

student = Student("Мира")
print(student.name)
```

### Слайд 6. Метод добавления товара

Вызов cart.add(250) меняет сумму этой корзины.

```python
class Cart:
    def __init__(self):
        self.total = 0

    def add(self, price):
        self.total += price
cart = Cart()
cart.add(250)
print(cart.total)
```

### Слайд 8. Две независимые корзины

Каждый вызов Cart() создаёт новую корзину со своей суммой.

```python
first = Cart()
second = Cart()
first.add(250)
print(first.total, second.total)
```

### Слайд 12. Метод с return

with_delivery возвращает итог. В корзине хранится сумма товаров.

```python
class Cart:
    def __init__(self, total):
        self.total = total
    def with_delivery(self, fee):
        return self.total + fee

cart = Cart(250)
print(cart.with_delivery(50), cart.total)
```

### Слайд 13. Метод clear

clear() присваивает текущей корзине сумму 0.

```python
class Cart:
    def __init__(self, total):
        self.total = total
    def clear(self):
        self.total = 0

cart = Cart(350)
cart.clear()
print(cart.total)
```

### Слайд 14. Список оценок у каждого студента

self.scores = [] в __init__ создаёт отдельный список.

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.scores = []

mira = Student("Мира")
aliya = Student("Алия")
mira.scores.append(85)
print(mira.scores, aliya.scores)
```
