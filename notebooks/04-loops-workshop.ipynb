# В какой день выполнена учебная цель?

Наставник дал студенту цель — посвятить упражнениям не менее 45 минут за несколько дней. Постройте накопительный итог и определите первый день, когда условие выполнено. Пустой по занятиям день всё равно присутствует в календаре.

Минуты практики по дням: 12, 0, 18, 9, 6, 20. Нумерация дней начинается с единицы. Длительности неотрицательные, цель положительная.

Выполняйте ячейки по порядку. Перед запуском попробуйте предсказать вывод.

## Шаг 1. Сколько минут в журнале и сколько из них пришлось на каждый день?

```python
minutes = [12, 0, 18, 9, 6, 20]
total = 0
day = 0
for duration in minutes:
    day += 1
    total += duration
    print(day, duration, total)
```

**Обсуждение.** Три числа в строке означают номер дня, продолжительность занятия и сумму с начала наблюдения. При значении 0 номер дня растёт, хотя сумма остаётся прежней. Поэтому вторую строку журнала важно сохранить в обходе.

## Шаг 2. Как остановиться на первом выполнении цели?

```python
goal = 45
day = 0
total = 0
for duration in minutes:
    day += 1
    total += duration
    if total >= goal:
        break
print(day, total)
```

**Обсуждение.** Перед новым обходом оба накопителя снова начинаются с нуля. Сравнение выполняется после добавления текущего дня, поэтому в проверку уже входят его шесть минут. При первом подходящем результате break прекращает цикл и сохраняет найденный номер.

## Шаг 3. Как различить день достижения и конец слишком короткого журнала?

```python
day = 0
total = 0
reached_day = None
for duration in minutes:
    day += 1
    total += duration
    if total >= goal:
        reached_day = day
        break
if reached_day is None:
    print("Цель пока не достигнута")
else:
    print("День достижения:", reached_day)
```

**Обсуждение.** Переменная reached_day изначально хранит None — отметку об отсутствии ответа. Номер попадает в неё только при выполнении условия. Теперь последний день нельзя случайно принять за успешный, а пустой журнал тоже даст осмысленное сообщение.

## Поиск логической ошибки
Почему результат может ввести в заблуждение?

```python
day = 0
total = 0
for duration in minutes:
    day += 1
    total += duration
    if total > goal:
        break
print(day, total)
```

Строгий знак > пропускает ровно 45 минут на пятом дне. На шестом дне условие выполняется, но ответ уже запоздал. Для цели «не менее» сравнение записывают как >=.
